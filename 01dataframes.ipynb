### DataFrame

```python
# dataframe : it is the tabular representation of the data

basic = pd.DataFrame([1, 2, 3, 4, 5], index=["a", "b", "c", "d", "e"], columns=["ok"])
print(basic)

details = {
    "name": ["raja", "Dhairya", "Raj"],
    "age": [12, 14, 15],
    "classes": [8, 9, 10],
}
details_dataframe =pd.DataFrame(details)
print(pd.DataFrame(details))
```
