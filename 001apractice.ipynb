### renaming the coloumn in an array 

```python
import pandas as pd

arr_12 = {
    "name": ["raja", "raj", "sakshi"],
    "class": ["1-A", "2-B", "3-C"],
    "roll": ["010", "045", "035"],
}
arr_1=pd.DataFrame(arr_12)
print(pd.DataFrame(arr_12))

# pd.DataFrame(arr_12).rename( columns ={"class":"class_section"})#this rename column only when this is run 

#to permanently change this code 
arr_1.rename( columns ={"class":"class_section"}, inplace =True)
print(arr_1)
```
