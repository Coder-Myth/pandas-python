### this keeps index from 1 to n in row as well coloumn

```python
import pandas as pd 
basic = pd.DataFrame([1,2,3,4,5,6,7,8])
print(basic)
```

### to rename the columns and the rows indexing value or name changing 
### use index and coloumn and rename them in the same dataframe bracket

```python
import pandas as pd
basic2 = pd.DataFrame([1,2,3,4,5], index =["a","b","c","d","e"], columns =["hello"])
print(basic2)
```

### renaming the coloumn in an array 

```python
import pandas as pd

arr_1 = {
    "name": ["raja", "raj", "sakshi"],
    "class": ["1-A", "2-B", "3-C"],
    "name": ["010", "045", "035"],
}
new_1=pd.DataFrame(arr_1)
print(pd.DataFrame(arr_1))

pd.DataFrame(arr_1).rename( columns ={"class":"class_section"})#this rename column only when this is run 

#to permanently change this code 
new_1.rename( columns ={"class":"class_section"}, inplace =True)
print(new_1)
```
