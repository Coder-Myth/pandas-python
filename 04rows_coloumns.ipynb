## Coloumns Selection

```python
import pandas as pd

details = {
    "name": ["raja", "Dhairya", "Raj"],
    "age": [12, 14, 15],
    "classes": [8, 9, 10],
}
details_dataframe = pd.DataFrame(details)
print(details_dataframe)
```

```python
details_dataframe[["name"]]#single coloumn selection 
details_dataframe[["name", "classes"]]#Multiple coloumn selection 
```

```python
# row selection
# loc  : label based indexing
details_dataframe.loc[
    (details_dataframe.name == "Dhairya") & (details_dataframe.classes > 5)
]
# used condition if the same person has is in the list


# code is made case sensitive to keep the practice safe and memorable
```

```python
#iloc : index label based indexing 
details_dataframe.iloc[0]

details_dataframe.loc[0:2]
```
