## Filter Dataframe:

```python
# syntax : filename["columnname"] (condition) ====> boolean value for each 
# syntax : filename[boolean value syntax] ====> only value that are true
# store them in a variable mostly in this

import pandas as pd
arr_12 = {
    "name": ["raja", "raj", "sakshi"],
    "class": ["1-A", "2-B", "3-C"],
    "roll": [10, 45, 35],
}
arr_1=pd.DataFrame(arr_12)

# returns boolean value 
arr_1['roll'] > 15


# returns value that are true 
arr_1[arr_1['roll']> 15] 

# using different condition in single dataframe
arr_1[(arr_1['roll']> 15) &(arr_1['name']=="raj")]

# where keyword
arr_1.where(arr_1['roll']>15)
    
```
