# rows and columns operation
### add new column
### syntax : filename['column_name']=[each index detail]

```python
import pandas as pd

arr_12 = {
    "name": ["raja", "raj", "sakshi"],
    "class": ["1-A", "2-B", "3-C"],
    "roll": ["010", "045", "035"],
}
arr_1=pd.DataFrame(arr_12) 


# column operation 
# new column added
arr_1['degree']=["be", "btech", "bca"]
arr_1

# operation with exsiting column
# column when integer then the same applies
arr_1["roll"]= arr_1["roll"]
arr_1

# Row operation
# this performs label based indexing 
#new row added
arr_1.loc[len(arr_1)]=["raja", "class", "014", "bba"]
arr_1

# Update Value:
arr_1.loc[1,"class"] = "3-b"
arr_1

# update by column only 
arr_1.loc[arr_1.name=='sakshi',"roll"] ="065"
arr_1

#delete value
# column drop 
arr_1.drop("class", axis =1)
arr_1.drop("class", axis =1)


#row drop
arr_1.drop(arr_1[arr_1.name=="Raja"].index , inplace=True)
arr_1


# Sort Values :
# this happens only in integer
# arr_1.sort_values["roll"]
```
